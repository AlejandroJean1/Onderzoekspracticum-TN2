```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
p_base = 4.88 #mW
p_objective = 3.49 #mW
p_loss_ratio = p_objective/p_base
measured_p = np.array([5.26,4.52,4.00,3.51,2.99,2.50,1.99,1.495,0.998,0.500,0.00297])
measured_I = np.array([16600,16200,15600,14300,12700,10900,9500,8300,7000,5200,3400])
P = measured_p*p_loss_ratio
def intensity(P,P_sat,I_max,b):
    I = I_max*(P/(P+P_sat)) +b
    return I
plt.plot(P,measured_I)
x_array = np.linspace(0,6,1000)
popt,cov = curve_fit(intensity,P,measured_I, bounds=([0,0,0], [50,500000,30000]))
print(popt)
plt.plot(x_array,intensity(x_array,*popt))
```
